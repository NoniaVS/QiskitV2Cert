```python
from qiskit import QuantumCircuit
from qiskit_ibm_runtime import SamplerV2 as Sampler, QiskitRuntimeService, fake_provider
from qiskit.transpiler import generate_preset_pass_manager
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(1,0)
qc.measure_all()

service = QiskitRuntimeService()
#backend = service.backend('ibm_basquecountry')
backend = fake_provider.FakeAuckland()
sampler = Sampler(mode = backend)

pm = generate_preset_pass_manager(optimization_level=1, backend=backend)
isa_circ = pm.run(qc)

pub = (isa_circ, None, 2000)
job = sampler.run([pub], shots=1000)
result = job.result()
```

```python
from qiskit import QuantumCircuit
from qiskit_ibm_runtime import SamplerV2 as Sampler, QiskitRuntimeService, fake_provider
from qiskit.transpiler import generate_preset_pass_manager

qc = QuantumCircuit(2)
qc.h(0)
qc.cx(1,0)
qc.measure_all()

service = QiskitRuntimeService()
#backend = service.backend('ibm_basquecountry')
backend = fake_provider.FakeAuckland()
sampler = Sampler(mode = backend)

pm = generate_preset_pass_manager(optimization_level=1, backend=backend)
isa_circ = pm.run(qc)

pub = (isa_circ)
sampler.options.default_shots=3000
job = sampler.run([pub])
result = job.result()
```
